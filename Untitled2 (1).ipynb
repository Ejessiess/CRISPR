```python
#过滤非对照扰动细胞数 <50
before = adata.n_obs
counts = adata.obs["perturbation"].value_counts()
non_ctrl_counts = counts[~counts.index.isin(CTRL_LABELS)]
small_perts = non_ctrl_counts[non_ctrl_counts < 50].index.tolist()

adata = adata[~adata.obs["perturbation"].isin(small_perts), :].copy()
print(f"[B] 去除非对照且细胞数 < 50 的扰动：{len(small_perts)} 个；保留细胞 {adata.n_obs}/{before}")
```

```python
#过滤无效扰动
#标准化
import numpy as np
import pandas as pd
import scanpy as sc
from scipy.sparse import issparse

CTRL_LABELS = {"control"}

def to_dense(x):
    return x.toarray() if issparse(x) else x

# 复制一份做轻量标准化（不会改原始 adata）
tmp = adata.copy()
sc.pp.normalize_total(tmp, target_sum=1e4)  # CPM/TP10K
sc.pp.log1p(tmp)

# 稀疏→稠密矩阵，方便后续 NumPy 运算
X = to_dense(tmp.X)
print("标准化矩阵形状:", X.shape)

# 对照组布尔掩码 & 平均表达向量
ctrl_mask = tmp.obs["perturbation"].isin(CTRL_LABELS).to_numpy()
assert ctrl_mask.sum() > 0, "找不到对照组，请检查 CTRL_LABELS 的名字"
ctrl_mean = X[ctrl_mask, :].mean(axis=0)
print("对照组细胞数:", int(ctrl_mask.sum()))

```

```python
#计算扰动效果
# 对每个扰动：均值表达向量 mu，与对照的差是logFC
# 计算两个指标：
#  median_abs_logFC：|logFC| 的中位数（整体变化强弱）
#  n_genes_abs_logFC_gt_0_25：|logFC| > 0.25 的基因数（明显变化基因数）
effect_stats = []
for pert in tmp.obs["perturbation"].unique():
    if pert in CTRL_LABELS:
        continue
    m = (tmp.obs["perturbation"] == pert).to_numpy()
    if m.sum() == 0:
        continue

    mu = X[m, :].mean(axis=0)     # 该扰动的平均表达
    logfc = mu - ctrl_mean        # 与对照的差异

    med_abs = float(np.median(np.abs(logfc)))
    n_large = int((np.abs(logfc) > 0.25).sum())  # 0.25 是经验阈值
    effect_stats.append({
        "perturbation": pert,
        "n_cells": int(m.sum()),
        "median_abs_logFC": med_abs,
        "n_genes_abs_logFC>0.25": n_large
    })

eff_df = pd.DataFrame(effect_stats).sort_values("median_abs_logFC")
print("扰动数（不含对照）:", len(eff_df))
eff_df.head(10)  # “最弱”的前 10 个扰动
```

```python
#找出无效扰动
INEFF_MED_ABS = 0.02   # 整体变化阈值：median |logFC| < 0.02 很弱
INEFF_N_LARGE  = 50    # 明显变化基因数阈值：< 50  很少

inefficient = eff_df[
    (eff_df["median_abs_logFC"] < INEFF_MED_ABS) &
    (eff_df["n_genes_abs_logFC>0.25"] < INEFF_N_LARGE)
]["perturbation"].tolist()

print(f"判定为“无效扰动”的个数: {len(inefficient)}")
eff_df[eff_df["perturbation"].isin(inefficient)].sort_values("median_abs_logFC").head(20)
```

```python
#删除无效扰动
before = adata.n_obs
adata = adata[~adata.obs["perturbation"].isin(inefficient)].copy()
print(f"去除无效扰动：{len(inefficient)} 个；细胞保留 {adata.n_obs}/{before}")
```

```python
#过滤低表达
import numpy as np

# 转成稠密矩阵便于计算
X_dense = adata.X.toarray() if hasattr(adata.X, "toarray") else adata.X

# 每个基因在多少个细胞中有非零表达
gene_counts = np.sum(X_dense > 0, axis=0)

print("基因总数:", X_dense.shape[1])
print("每个基因在多少细胞中表达:", gene_counts[:10])  # 看前10个
```

```python
#过滤掉
MIN_CELLS_EXPRESSED = 100  # 少于100个细胞表达的基因将被过滤

mask = gene_counts >= MIN_CELLS_EXPRESSED
before = adata.n_vars  # n_vars = 基因数
adata = adata[:, mask].copy()
after = adata.n_vars

print(f"[D] 基因过滤完成：{after}/{before} 保留 (>= {MIN_CELLS_EXPRESSED} 细胞表达)")
```

```python
#构建Preudobulk矩阵
import pandas as pd
import numpy as np

group_keys = ["perturbation", "celltype"]

# 转为稠密矩阵
X_dense = adata.X.toarray() if hasattr(adata.X, "toarray") else adata.X

# 把表达矩阵转换为 DataFrame，与 adata.obs 一一对应
expr_df = pd.DataFrame(X_dense, index=adata.obs_names, columns=adata.var_names)
expr_df[group_keys] = adata.obs[group_keys].values

# 计算每个组的平均表达（pseudobulk）
pseudobulk_df = expr_df.groupby(group_keys).mean()

print("伪bulk矩阵形状：", pseudobulk_df.shape)
pseudobulk_df.head()
```

```python
#PCA
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 先对 pseudobulk 数据做标准化（每个基因 z-score）
X_scaled = (pseudobulk_df - pseudobulk_df.mean(axis=0)) / pseudobulk_df.std(axis=0)

# PCA 拟合
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

# 画图
plt.figure(figsize=(6, 5))
for pert in pseudobulk_df.index.get_level_values("perturbation").unique():
    idx = pseudobulk_df.index.get_level_values("perturbation") == pert
    plt.scatter(pca_result[idx, 0], pca_result[idx, 1], label=pert, s=80)
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA on Pseudobulk Expression")
plt.show()

print("PCA 方差解释率:", pca.explained_variance_ratio_)
```

```python
#差异分析
import scanpy as sc
import pandas as pd
import numpy as np

# cell type---K562，并保留 AARS / control 两组
celltype_to_use = "K562"
adata_sub = adata[(adata.obs["celltype"] == celltype_to_use) &
                  (adata.obs["perturbation"].isin(["AARS", "control"]))].copy()

print(adata_sub.shape, adata_sub.obs["perturbation"].value_counts())

sc.pp.normalize_total(adata_sub, target_sum=1e4)
sc.pp.log1p(adata_sub)

# 以“perturbation”为分组做差异分析（AARS 相对 control）
sc.tl.rank_genes_groups(
    adata_sub,
    groupby="perturbation",
    reference="control",     #
    method="wilcoxon",
    use_raw=False
)

# 可视化（前 20 个基因）
sc.pl.rank_genes_groups(adata_sub, n_genes=20, sharey=False)

# 结果表
res_df = sc.get.rank_genes_groups_df(adata_sub, group="AARS")
# res_df 包含 names, scores, logfoldchanges, pvals, pvals_adj 等
res_df.head(10)
```

```python
#火山图
import matplotlib.pyplot as plt

# 绘制火山图
plt.figure(figsize=(6,6))
plt.scatter(res_df['logfoldchanges'], -np.log10(res_df['pvals_adj']),
            c=(res_df['pvals_adj']<0.05), cmap='coolwarm', alpha=0.7)
plt.axvline(x=0.25, color='gray', linestyle='--')
plt.axvline(x=-0.25, color='gray', linestyle='--')
plt.axhline(y=-np.log10(0.05), color='gray', linestyle='--')
plt.xlabel('log2 Fold Change')
plt.ylabel('-log10(FDR)')
plt.title('AARS vs Control')
plt.show()

# 导出显著差异基因表
sig_df = res_df.query("pvals_adj < 0.05 and abs(logfoldchanges) > 0.25")
sig_df.to_csv("AARS_vs_Control_DEG.csv", index=False)
print(f"显著差异基因数：{sig_df.shape[0]}")
```
